```python
plt.style.use("dark_background")
results = {10: 1.5953, 50: 1.5983, 100: 1.5932}

points = list(results.keys())
losses = list(results.values())

plt.figure(figsize=(8,5))

plt.plot(points, losses, marker='o', linewidth=3, color="#00E5FF")

plt.title("PFCE Experiment Results")
plt.xlabel("Points per Image")
plt.ylabel("Validation Loss")

plt.show()
```

```python
plt.style.use("dark_background")
plt.figure(figsize=(8,5))

colors = ['#FF6B6B', '#4ECDC4', '#1A535C']

plt.bar(points, losses, color=colors)

plt.title("Validation Loss vs Number of Points", fontsize=14)
plt.xlabel("Number of Points")
plt.ylabel("Validation Loss")

plt.show()
```

```python
plt.style.use("dark_background")
plt.figure(figsize=(8,5))

sns.lineplot(x=points, y=losses, marker='o', linewidth=2.5)

plt.title("PFCE Experiment Results", fontsize=16)
plt.xlabel("Points per Image")
plt.ylabel("Validation Loss")

plt.show()
```
